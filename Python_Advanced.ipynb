```python
# Task 14: Restaurant Ordering System

menu= {
    "Burger":500,
    "Pizza":1200,
    "Pasta":800,
    "Drink":200
}

orders = []

print("========= Restaurant Menu =========")
for key,value in menu.items():
  print(f'{key} : {value}')

while True:
  item_name = input("Enter your item name: ").title()
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
    print("This item is not available")

  choice = input("You want to add more items: yes/no: ")
  if choice == "no":
    break

total = 0
for order in orders:
  total = total + order["sub_total"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1
else:
  discount = 0

grand_total = total + tax - discount

def receipt():
  print("\n============ Receipt ============")

  for order in orders:
    print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')

  print("Total: ", total)
  print("Tax: ", tax)
  print("Discount: ", discount)
  print("Grand Total: ", grand_total)

receipt()
```

```python
# Task 15: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient found")
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')
      print(f'Patient emergency: {patient["emergency"]}')
      return
    else:
      print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("Invalid choice")
```

```python
#Task 10
expenses = [
    {
        "name": "Biryani",
        "category": "Food",
        "amount": 500
    },
    {
        "name": "Bus",
        "category": "Transport",
        "amount": 200
    },
    {
        "name": "Shoes",
        "category": "Shopping",
        "amount": 1000
    }
]
categories = ["Food", "Transport", "Shopping", "Bills", "Other"]

# Total number of expenses
n = int(input("Enter total number of expenses: "))

# Taking expense details
for i in range(n):
    print("\nExpense", i + 1)

    name = input("Enter expense name: ")

    print("Categories:", categories)
    category = input("Enter category: ")

    while category not in categories:
        print("Invalid category!")
        category = input("Enter category again: ")

    amount = float(input("Enter amount: "))

    # Store expense in dictionary
    expense = {
        "name": name,
        "category": category,
        "amount": amount
    }

    # Store dictionary in list
    expenses.append(expense)


# Budget input
budget = float(input("\nEnter your budget: "))


# Function to display final report
def final_report(expenses, budget):

    # Calculate total spending
    total = 0

    for expense in expenses:
        total += expense["amount"]

    # Calculate category-wise spending
    category_total = {
        "Food": 0,
        "Transport": 0,
        "Shopping": 0,
        "Bills": 0,
        "Other": 0
    }

    for expense in expenses:
        category = expense["category"]
        category_total[category] += expense["amount"]

    # Find highest expense
    highest = expenses[0]

    for expense in expenses:
        if expense["amount"] > highest["amount"]:
            highest = expense

    # Display report
    print("\n========== EXPENSE REPORT ==========")

    print("Total Spending:", total)

    print("\nCategory-wise Spending:")

    for category in category_total:
        print(category, ":", category_total[category])

    print("\nHighest Expense:")
    print("Name:", highest["name"])
    print("Category:", highest["category"])
    print("Amount:", highest["amount"])

    # Check budget
    if total > budget:
        print("\nWARNING: Budget Exceeded!")
        print("Extra Spending:", total - budget)
    else:
        print("\nBudget is not exceeded.")
        print("Remaining Budget:", budget - total)


# Call the function
final_report(expenses, budget)
```

```python
#Task 11
products = {
    "Pen": {
        "price": 50,
        "quantity": 20
    },

    "Notebook": {
        "price": 200,
        "quantity": 10
    },

    "Pencil": {
        "price": 20,
        "quantity": 50
    }
}
# 1 & 2. New Product Add Function
def add_product():
    name = input("Enter product name: ")
    price = float(input("Enter product price: "))
    quantity = int(input("Enter quantity: "))

    products[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully!")


# 3 & 4. Sell Product
def sell_product():
    name = input("Enter product name: ")

    if name in products:
        quantity = int(input("Enter quantity to sell: "))

        if quantity <= products[name]["quantity"]:
            products[name]["quantity"] -= quantity
            print("Product sold successfully!")
        else:
            print("Warning: Insufficient stock!")

    else:
        print("Product not found!")


# 5. Restock Product
def restock_product():
    name = input("Enter product name: ")

    if name in products:
        quantity = int(input("Enter quantity to add: "))

        products[name]["quantity"] += quantity

        print("Product restocked successfully!")

    else:
        print("Product not found!")


# 6. Search Product
def search_product():
    name = input("Enter product name to search: ")

    if name in products:
        print("\nProduct Found!")
        print("Name:", name)
        print("Price:", products[name]["price"])
        print("Quantity:", products[name]["quantity"])

    else:
        print("Product not found!")


# 7 & 8. Display Low Stock Products
def low_stock():
    print("\nLow Stock Products:")

    found = False

    for name, details in products.items():
        if details["quantity"] < 5:
            print(
                name,
                "- Quantity:",
                details["quantity"]
            )
            found = True

    if not found:
        print("No low-stock products.")


# Display Inventory
def display_inventory():
    print("\n========== INVENTORY ==========")

    if len(products) == 0:
        print("Inventory is empty.")
        return

    for name, details in products.items():
        print("Product:", name)
        print("Price:", details["price"])
        print("Quantity:", details["quantity"])
        print("------------------------")


# 9. Calculate Total Inventory Value
def total_inventory_value():
    total = 0

    for name, details in products.items():
        product_value = details["price"] * details["quantity"]
        total += product_value

    print("\nTotal Inventory Value:", total)


# Main Menu
while True:

    print("\n========== SHOP MENU ==========")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Display Low Stock")
    print("7. Total Inventory Value")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        low_stock()

    elif choice == "7":
        total_inventory_value()

    elif choice == "8":
        print("Program ended.")
        break

    else:
        print("Invalid choice! Please try again.")


```

```python
#Task 13
books = [
    {
        "id": "B001",
        "title": "Python Programming",
        "author": "John Smith",
        "available": True
    },
    {
        "id": "B002",
        "title": "Data Structures",
        "author": "Robert Brown",
        "available": True
    },
    {
        "id": "B003",
        "title": "Introduction to AI",
        "author": "Sarah Khan",
        "available": False
    },
    {
        "id": "B004",
        "title": "Web Development",
        "author": "Ali Ahmed",
        "available": True
    }
]
# 1. Add Book
def add_book():
    book_id = input("Enter Book ID: ")

    # Check unique Book ID
    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book added successfully!")


# 2. Display Books
def display_books():
    if len(books) == 0:
        print("No books available.")
        return

    print("\n========== BOOKS ==========")

    for book in books:
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])

        if book["available"]:
            print("Status: Available")
        else:
            print("Status: Borrowed")

        print("------------------------")


# 3. Search Book by Title
def search_book():
    title = input("Enter book title to search: ")

    found = False

    for book in books:
        if book["title"].lower() == title.lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            found = True

    if not found:
        print("Book not found!")


# 4. Borrow Book
def borrow_book():
    book_id = input("Enter Book ID to borrow: ")

    for book in books:
        if book["id"] == book_id:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")

            return

    print("Book not found!")


# 5. Return Book
def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in books:
        if book["id"] == book_id:

            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("Book is already available!")

            return

    print("Book not found!")


# Main Menu
while True:

    print("\n========== LIBRARY MENU ==========")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice!")
```
