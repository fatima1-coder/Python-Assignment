```python
#Task 1
name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age < 5 or age > 100:
    price = 0
elif age >= 5 and age <= 12:
    price = 300
elif age >= 13 and age <= 59:
    price = 700
elif age >= 60 and age <= 100:
    price = 400

print("Customer Name:", name)
print("Ticket Price: Rs.", price)
```

```python
#Task 2
celsius = float(input("Enter Celsius temperature: "))

fahrenheit = (celsius * 9 / 5) + 32

if celsius < 10:
    print("Weather: Very Cold")
elif celsius >= 10 and celsius <= 19:
    print("Weather: Cold")
elif celsius >= 20 and celsius <= 29:
    print("Weather: Normal")
else:
    print("Weather: Hot")

print("Fahrenheit:", fahrenheit)
```

```python
#Task 3
mobile_number = input("Enter mobile number: ")
recharge_amount = float(input("Enter recharge amount: "))

tax = recharge_amount * 3 / 100
final_balance = recharge_amount - tax

print("Mobile Number:", mobile_number)
print("Recharge Amount: Rs.", recharge_amount)
print("Tax (3%): Rs.", tax)
print("Final Received Balance: Rs.", final_balance)
```

```python
#Task 4
total_steps = 0

for day in range(1, 8):
    steps = int(input(f"Enter steps for Day {day}: "))
    total_steps += steps

average_steps = total_steps / 7

print("Total Steps:", total_steps)
print("Daily Average:", average_steps)

if average_steps >= 5000:
    print("Active Week")
else:
    print("Needs More Activity")
```

```python
#Task 5
secret_pin = "1234"
attempts = 3

while attempts > 0:
    pin = input("Enter your PIN: ")

    if pin == secret_pin:
        print("Access Granted")
        break
    else:
        attempts -= 1

        if attempts > 0:
            print("Incorrect PIN")
            print("Remaining attempts:", attempts)
        else:
            print("Account Blocked")


```
